1 y 2. GET a posts, users y todos — cantidad total y código de estado

```python
import requests

BASE = "https://jsonplaceholder.typicode.com"

posts = requests.get(f"{BASE}/posts")
users = requests.get(f"{BASE}/users")
todos = requests.get(f"{BASE}/todos")

print(f"Posts: {len(posts.json())} registros | Status: {posts.status_code}")
print(f"Users: {len(users.json())} registros | Status: {users.status_code}")
print(f"Todos: {len(todos.json())} registros | Status: {todos.status_code}")
```

3. Petición a publicación inexistente (error 404)

```python
respuesta_404 = requests.get(f"{BASE}/posts/9999")
print(f"Status recibido: {respuesta_404.status_code}")
```

4. POST — Crear nueva publicación

```python
nueva_publicacion = {
    "title": "Mi nueva publicación",
    "body": "Este es el cuerpo del mensaje de prueba.",
    "userId": 1
}

respuesta_post = requests.post(f"{BASE}/posts", json=nueva_publicacion)

print("Respuesta JSON:", respuesta_post.json())
print(f"Status: {respuesta_post.status_code}")
```

5. PATCH — Modificar parcialmente una publicación existente

```python
modificacion = {"title": "Título modificado con PATCH"}

respuesta_patch = requests.patch(f"{BASE}/posts/1", json=modificacion)

print("Respuesta JSON:", respuesta_patch.json())
print(f"Status: {respuesta_patch.status_code}")
```

6. DELETE — Eliminar una publicación

```python
respuesta_delete = requests.delete(f"{BASE}/posts/1")

print("Respuesta JSON:", respuesta_delete.json())
print(f"Status: {respuesta_delete.status_code}")
```

Nivel 2 — REST Countries API

Documentación de la API

**API elegida:** [REST Countries](https://restcountries.com/) — API pública y gratuita, sin autenticación. Respuestas en formato JSON.

---

Endpoints disponibles

| Endpoint | Descripción |
|---|---|
| `GET /v3.1/all` | Devuelve todos los países del mundo |
| `GET /v3.1/region/{region}` | Filtra países por región (Europe, Asia, Americas, Africa, Oceania) |
| `GET /v3.1/name/{name}` | Busca un país por nombre |
| `GET /v3.1/lang/{language}` | Filtra países por idioma oficial |

---

Parámetros opcionales interesantes

- `fields`: selecciona únicamente los campos deseados para reducir el tamaño de la respuesta.
  Ejemplo: `?fields=name,capital,population,region,area`

3. Petición GET — países de Europa con campos seleccionados

```python
import pandas as pd

URL_COUNTRIES = "https://restcountries.com/v3.1/region/europe"
params = {"fields": "name,capital,population,area"}

respuesta = requests.get(URL_COUNTRIES, params=params)
print(f"Status: {respuesta.status_code}")

datos = respuesta.json()
print(f"Total de países en Europa: {len(datos)}\n")

for pais in datos[:5]:
    nombre    = pais["name"]["common"]
    capital   = pais.get("capital", ["N/A"])[0]
    poblacion = pais.get("population", 0)
    area      = pais.get("area", 0)
    print(f"  {nombre:<25} | Capital: {capital:<20} | Población: {poblacion:>12,.0f} | Área km²: {area:>12,.0f}")
```

4. Convertir a DataFrame

```python
filas = []
for pais in datos:
    filas.append({
        "pais":      pais["name"]["common"],
        "capital":   pais.get("capital", [None])[0],
        "poblacion": pais.get("population"),
        "area_km2":  pais.get("area")
    })

df_europa = pd.DataFrame(filas).sort_values("poblacion", ascending=False).reset_index(drop=True)
df_europa.head(10)
```

Nivel 3 — API Open Data BCN

1. Búsqueda de dataset con `package_search`

```python
BASE_BCN = "https://opendata-ajuntament.barcelona.cat/data/api/3/action"

params_search = {"q": "accidents", "rows": 5}
respuesta_search = requests.get(f"{BASE_BCN}/package_search", params=params_search)

print(f"Status: {respuesta_search.status_code}")
resultados = respuesta_search.json()["result"]["results"]
print(f"Datasets encontrados: {len(resultados)}\n")

for ds in resultados:
    print(f"  ID:     {ds['id']}")
    print(f"  Nombre: {ds['name']}")
    print(f"  Título: {ds.get('title', 'N/A')}")
    print()
```

1b. Detalle del dataset con `package_show`

```python
dataset_id = resultados[0]["name"]
print(f"Dataset seleccionado: {dataset_id}\n")

respuesta_show = requests.get(f"{BASE_BCN}/package_show", params={"id": dataset_id})
detalle = respuesta_show.json()["result"]

print(f"Título: {detalle.get('title', 'N/A')}")
print(f"Notas:  {str(detalle.get('notes', ''))[:200]}\n")
print("Recursos disponibles:")
for rec in detalle.get("resources", []):
    print(f"  - {rec['name']} | Formato: {rec['format']} | Datastore activo: {rec.get('datastore_active', False)} | ID: {rec['id']}")
```

3. Consulta con `datastore_search` (mínimo 100 registros)

```python
resource_id = None
for rec in detalle.get("resources", []):
    if rec.get("datastore_active"):
        resource_id = rec["id"]
        print(f"Recurso con datastore activo: {rec['name']} (ID: {resource_id})")
        break

if resource_id is None:
    resource_id = detalle["resources"][0]["id"]
    print(f"Usando primer recurso disponible (ID: {resource_id})")

respuesta_ds = requests.get(
    f"{BASE_BCN}/datastore_search",
    params={"resource_id": resource_id, "limit": 100}
)

print(f"\nStatus: {respuesta_ds.status_code}")
registros = respuesta_ds.json()["result"]["records"]
print(f"Registros obtenidos: {len(registros)}")
```

4. Convertir a DataFrame

```python
df_bcn = pd.DataFrame(registros)
print(f"Shape: {df_bcn.shape}")
df_bcn.head()
```

5. Guardar en archivo CSV

```python
ruta_csv = "accidents_barcelona.csv"
df_bcn.to_csv(ruta_csv, index=False, encoding="utf-8-sig")
print(f"Archivo guardado: {ruta_csv}")
print(f"Filas: {len(df_bcn)} | Columnas: {len(df_bcn.columns)}")
```
