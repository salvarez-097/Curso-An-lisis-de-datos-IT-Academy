- Escribe una función que calcule el IMC ingresado por el usuario/a, es decir, quien lo ejecute deberá ingresar estos datos. Puedes obtener más información de su cálculo en: **[Índice de masa corporal IMC que es y cómo se calcula](https://muysalud.com/salud/indice-de-masa-corporal-imc-que-es-y-como-se-calcula/)**.

- La función debe clasificar el resultado en sus respectivas categorías.

Consejo: Intenta validar los datos previamente, para que envíe un mensaje de advertencia si los datos introducidos por el usuario/a no están en el formato adecuado o no toma valores razonables.

```python
import unicodedata

print("Bienvenido/a a la calculadora de IMC\n")

# Esta funcion recibe los parametros de peso y altura, calcula la altura al cuadrado, calcula el IMC y ejecuta la condicion para
# mostrar que IMC se imprime en funcion del resultado obtenido al calcular el IMC
def calcular_IMC(peso, altura):
    
    altura_al_cuadrado = altura * altura

    imc = peso / altura_al_cuadrado
    imc = round(imc, 2)

    if imc >= 40:
        return(f"Su IMC es {imc}, usted tiene obesidad tipo III.\n")

    elif imc >= 35:
        return(f"su IMC es {imc}, usted tiene obesidad tipo II.\n")

    elif imc >= 30:
        return(f"Su IMC es {imc}, usted tiene obesidad tipo I.\n")

    elif imc >= 25:
        return(f"Su IMC es {imc}, usted tiene sobrepeso.\n")

    elif imc >= 18.5:
        return(f"Su IMC es {imc}, usted tiene peso normal.\n")

    else:
        return(f"Su IMC es {imc}, usted tiene bajo peso.\n") 

# Esta funcion recoge los parametros introducidos por el usuario y valida que los datos introducidos son correctos
def parametros ():
    
    while True:
        try: 
            peso = float(input("Introduzca su peso en kg (Ej.: 25 o 73.5): "))
            
            if 2 <= peso <= 400:
                break
            else:
                print("Introduzca un peso real.")

        except ValueError:
            print("Introduzca un numero, no estan permitidas palabras.")

    while True:
        try:
            opcion = input("¿Prefiere introducir su altura en metros o centimetros?").lower()

            if opcion == "metros":
                altura = float(input("Introduzca su altura en metros (Ej.: 1.64 o 1.86). RECUERDA QUE ESTAS INTRODUCIENDO METROS, NO CENTIMETROS"))      
                if 0.5 <= altura <= 2.5:
                    break
                elif 100 <= altura <= 250:
                    print("Ha introducido su altura en centimetros, no se preocupes, se corrige automaticamente.")
                    altura = altura / 100
                    break
                else:
                    print("Altura no realista, introduzca una altura realista.") 
                        
            elif opcion == "centimetros":
                altura = float(input("Introduzca su altura en centimetros (Ej.: 164 o 186)"))
                if 50 <= altura <= 250:
                    altura = altura / 100
                    break
                else:
                    print("Introduzca una altura real.")
            else:
                print("Escriba metros o centimetros.")

        except ValueError:
            print("Debe introducir un número.")

    return peso, altura

# Función orquestadora, muestra el resultado y permite repetir calcular otro IMC
def main():
    while True:
    
        peso, altura = parametros()
        resultado = calcular_IMC(peso, altura)
        print(f"Su peso es {peso} kg\nSu altura es {altura} m\n{resultado}")

        repetir = input("¿Quiere calcular otro IMC? Si/No\n")
        
        repetir = unicodedata.normalize('NFKD', repetir)
        repetir = ''.join(c for c in repetir if not unicodedata.combining(c))
        repetir = repetir.lower()

        if repetir != "si":
            print("Hasta la proxima")
            break

main()
```

<h4>2. Segunda opción (celsius como pivote)</h4>

```python
# En esta funcion estan las conversiones con celsius como intermediario
def conversiones():
    a_celsius = {
        "fahrenheit": lambda f: (f - 32) * 5/9,
        "kelvin":     lambda k: k - 273.15,
        "rankine":    lambda r: (r - 491.67) * 5/9,
        "reaumur":    lambda re: re * 5/4,
        "celsius":    lambda c: c,
        }
    de_celsius = {
        "fahrenheit": lambda c: c * 9/5 + 32,
        "kelvin":     lambda c: c + 273.15,
        "rankine":    lambda c: (c + 273.15) * 9/5,
        "reaumur":    lambda c: c * 4/5,
        "celsius":    lambda c: c,
        }
    return a_celsius, de_celsius

# Esta función convierte los datos introdcidos por el usuario teniendo en cuenta las conversiones anteriores
def convertir(valor, origen, destino, a_celsius, de_celsius):
    en_celsius = a_celsius[origen](valor)
    resultado  = de_celsius[destino](en_celsius)
    return resultado

# La siguiente funcion recibe los datos introducidos por el usuario, con las validaciones necesarias para evitar que se
# inrtoduzcan datos erroneos.
def solicitud_datos():    
    while True:

        unidades_validas = ["celsius", "fahrenheit", "kelvin", "rankine", "reaumur"]
        while True:
            origen = input("¿Desde que temperatura quieres convertir?").strip().lower()

            if origen in unidades_validas:
                break
            else:
                print("Introduce una unidad valida de las siguientes: Celsius, Fharenheit, Kelvin, Rankine, Reaumur")
        
        while True:
            destino = input("¿A que temperatura quieres convertir?").strip().lower()

            if destino in unidades_validas:
                break
            else:
                print("Introduce una unidad valida de las siguientes: Celsius, Fharenheit, Kelvin, Rankine, Reaumur")
        break

    for nombre in unidades_validas:
        while True:
            try:
                while True:
                    try:
                        valor = float(input(f"¿Que temperatura quieres convertir?: "))

                        if nombre == "celsius" and valor < -273.15:
                            print("La temperatura inferior a -273.15 en Celsius es no existe.")
                            continue

                        elif nombre == "fahrenheit" and valor < -459.67:
                            print("Una temperatura menor a -459.67 en Fahrenheit no es real.")
                            continue

                        elif nombre == "kelvin" and valor < 0:
                            print("La temperatura negativa en Kelvin es fisicamente imposible.")
                            continue

                        elif nombre == "rankine" and valor < 0:
                            print("La temperatura negativa en Rankine no existe.")
                            continue

                        elif nombre == "reaumur" and valor < -218.52:
                            print("La temperatura menor a -218.52 en Reaumur no es posible.")
                            continue
                        break

                    except ValueError:
                        print("No se admiten letras, palabras o vacio.")
                break
            except ValueError:
                print("No se admiten letras, palabras o vacio.")
        break
    return valor, origen, destino
    
def main():
    datos = solicitud_datos()
    valor, origen, destino = datos

    conversion_datos = conversiones()
    a_celsius, de_celsius = conversion_datos

    transformacion = convertir(valor, origen, destino, a_celsius, de_celsius)
    transformacion = round(transformacion, 2)

    abreviatura_medidas = {"celsius":"ºC", "fahrenheit":"ºF", "kelvin":"K", "rankine":"ºRa", "reaumur":"ºRe"}

    entrada = origen
    salida = destino

    if entrada in abreviatura_medidas:
        abreviatura_entrada = abreviatura_medidas[entrada]

    if salida in abreviatura_medidas:
        abreviatura_salida = abreviatura_medidas[salida]

    #Diferentes formas de imprimir el resultado
    print(f"Los grados introducidos ({valor}) de {origen.capitalize()} a {destino.capitalize()} son {transformacion}{abreviatura_salida}")
    print(f"{valor} grados {origen.capitalize()} son {transformacion} grados {destino.capitalize()}")
    print(f"{valor}{abreviatura_entrada} son {transformacion}{abreviatura_salida}")

main()
    
```

<h3> Opcion mas óptima </h3>

```python
import string
import random

# Diccionario converiones, minimos y simbolos para visualización
A_CELSIUS = {
    "celsius":    lambda c: c,
    "fahrenheit": lambda f: (f - 32) * 5 / 9,
    "kelvin":     lambda k: k - 273.15,
    "rankine":    lambda r: (r - 491.67) * 5 / 9,
    "reaumur":    lambda re: re * 5 / 4,
}

DE_CELSIUS = {
    "celsius":    lambda c: c,
    "fahrenheit": lambda c: c * 9 / 5 + 32,
    "kelvin":     lambda c: c + 273.15,
    "rankine":    lambda c: (c + 273.15) * 9 / 5,
    "reaumur":    lambda c: c * 4 / 5,
}

MINIMOS = {
    "celsius":    -273.15,
    "fahrenheit": -459.67,
    "kelvin":     0,
    "rankine":    0,
    "reaumur":    -218.52,
}

SIMBOLOS = {
    "celsius":    "°C",
    "fahrenheit": "°F",
    "kelvin":     "K",
    "rankine":    "°Ra",
    "reaumur":    "°Ré",
}

# Funcion que comprueba si la unidad introducida existe en el conversor.
def unidad_valida(unidad):
    
    return unidad in A_CELSIUS

# Funcion que comprueba que el valor no esté por debajo del cero absoluto de la unidad seleccionada.
def temperatura_valida(valor, unidad):
    
    return valor >= MINIMOS[unidad]


# Funcion de conversion pasando por celsius
def convertir(valor, origen, destino):
   
    valor_celsius = A_CELSIUS[origen](valor)
    return round(DE_CELSIUS[destino](valor_celsius), 2)

# Funcion que convierte la temperatura seleccionada a todas las demás
def convertir_a_todas(valor, origen):

    valor_celsius = A_CELSIUS[origen](valor)
    return {
        destino: round(funcion(valor_celsius), 2)
        for destino, funcion in DE_CELSIUS.items()
        if destino != origen
    }


# Funcion para solicitar la medida desde la que se va a convertir
def pedir_unidad():

    opciones = ", ".join(A_CELSIUS.keys())
    while True:
        unidad = input(f"¿Desde qué unidad quieres convertir? ({opciones}): ").strip().lower()
        if unidad_valida(unidad):
            return unidad
        print(f"Unidad no valida. Elige entre: {opciones}.")

# Funcion para solicitar la temperatura a convertir
def pedir_temperatura(unidad):

    while True:
        try:
            valor = float(input(f"Introduce la temperatura en {unidad.capitalize()}: "))
            if temperatura_valida(valor, unidad):
                return valor
            print(f"Temperatura imposible. El mínimo para {unidad.capitalize()} es {MINIMOS[unidad]}.")
        except ValueError:
            print("Debes introducir un número válido.")


# Funcion orquestadora
def main():
    origen = pedir_unidad()
    valor  = pedir_temperatura(origen)

    resultados = convertir_a_todas(valor, origen)

    print(f"\nConversiones para {valor}{SIMBOLOS[origen]}:\n")
    
    for destino, resultado in resultados.items():
        print(f"  {destino.capitalize():<12} {resultado}{SIMBOLOS[destino]}")


main()
```

(EXTRA): ¿Cuál es la longitud media de las palabras del texto que has escrito? “¿Hola cómo va?” debería devolver (4+3+2) / 3 = 3

```python
import string
import re
 
# Esta funcion cuenta las palabras del texto introducido por el usuario
def contador_veces_palabra(texto):
    contador = {}
    for palabra in texto.split():
        if palabra in contador:
            contador[palabra] = contador[palabra] + 1
        else:
            contador[palabra] = 1
    return contador

# Esta funcion cuenta cuantas veces ha salido cada palabra, y se eliminan todos los caracteres que no formen parte de las palabras
def numero_letras_por_palabra(texto):    
    palabras = []
    for palabra in texto.split():
        palabras.append(len(palabra))
    return palabras

# Esta funcion calcula la longitud media de todas las palabras introducidas
def longitud_media_palabras(palabras):    
    suma = 0
    for num in palabras:
        suma = suma + num
    longitud_media = suma / len(palabras)
    return longitud_media

def limpiar_texto(texto):
   signos = string.punctuation + "¿¡"
   tabla = str.maketrans("", "", signos)
   limpio = re.sub(r'[^a-zA-Z0-9\s]', '', texto)
   return limpio.lower().translate(tabla)

# El codigo siguiente recibe el texto introducido por el usuario, con la validacion de que no se introduzcan 
# numeros, se eliminan los espacios al principio y al final y se imprime el resultado de las palabras y la 
# longitud media
def recibe_texto():
    while True:
        texto = input("Introduce un texto para contar las palabras:")
        
        if any(letra.isdigit() for letra in texto):
            print("No se admiten numeros.")
        else:
            break
    texto = texto.strip()
    return texto

def main():
    texto = recibe_texto()
    texto_limpio = limpiar_texto(texto)
    recuento = contador_veces_palabra(texto_limpio)
    palabras = numero_letras_por_palabra(texto_limpio)
    media = longitud_media_palabras(palabras)
    
    print(f"El texto original es: \"{texto}\"\n")
    
    for palabra in recuento:
        print(f"La palabra \"{palabra}\" se repite {recuento[palabra]} {'vez' if recuento[palabra] <= 1 else 'veces'}")
        print()
    
    print(f"La longitud media de las palabras introducidas es: {round(media, 2)}")

main()
```

```python
# Esta funcion invierte el diccionario de la misma manera que la anterior, 
# con la condicion que evalua y advierte de si un valor esta repetido o no, en este caso las calves no pueden estar repetidas
# porque en un diccionario no se pueden poner claves repetidas, son unicas; los valores repetidos los pone en la misma clave pero
# en forma de lista para que esten todos los datos del diccionario original.
def inversor(encuesta_antigua):
    encuesta_nueva = {}
    for clave,valor in encuesta_antigua.items():
        if valor in encuesta_nueva:
            print(f"Advertencia, el valor {valor} esta repetido.")
            if not isinstance(encuesta_nueva[valor], list):
                encuesta_nueva[valor] = [encuesta_nueva[valor]]
            encuesta_nueva[valor].append(clave)
        else:
            encuesta_nueva[valor] = clave
            
    return encuesta_nueva
    
encuesta_antigua = {'rojo':8, 'azul':4, 'naranja': 4, 'verde':14, 'blanco':7, 'negro':19, 'marron':5, 'lila':7, 'rosa':4, 'amarillo': 4}

inversion = inversor(encuesta_antigua)
print(inversion)
```

```python
lista = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

# Esta funcion es la orquestadora, la cual recibe la lista
def main(lista):  

    # Esta funcion transforma la lista original en dos listas, una tiene los elementos que se pueden convertir a flotantes y la otra lista
    # los que no pueden convertirse a flotantes, tambien tiene la validacion necesaria para que no se introduzcan datos erroneos, por 
    # ultimo se mete ambas listas en una tupla y se asigna a una variable. 
    def transformacion(lista):
        lista_conversiones_flotante = []
        lista_no_conversiones_flotante = []

        for elemento in lista:
            try:
                lista_conversiones_flotante.append(float(elemento))
            except TypeError:
                for subelemento in elemento:
                    try:
                        lista_conversiones_flotante.append(float(subelemento))
                    except (ValueError, TypeError):
                        lista_no_conversiones_flotante.append(subelemento)
            except ValueError:
                lista_no_conversiones_flotante.append(elemento)

        lista_nueva = (lista_conversiones_flotante, lista_no_conversiones_flotante)

        print(lista_nueva)
    
    transformacion(lista)

main(lista)
```

<h2>1. Generador de Contraseñas</h2>

Explora el funcionamiento del módulo random de la librería numpy. 🔗 Random module in NumPy

En este punto, el cliente ha detectado un problema con las contraseñas que utilizan sus trabajadores. Asdf1234, fechas de cumpleaños o similares. Para solucionarlo, nos ha encargado una función de Python que genere contraseñas más seguras. La función debe depender de los siguientes parámetros:

- longitud (int): Longitud de la contraseña

- mayúsculas (bool = True): Si tiene que aparecer mayúsculas

- minúsculas (bool = True): Si tiene que aparecer minúsculas

- números (bool = True): Si tiene que aparecer números

- signos (bool = False): Si tiene que aparecer caracteres especiales (,-$? o similares)

Así pues, si ejecutamos la función de la siguiente forma:

crear_contraseña(10, True, True, True, True)

Deberíamos obtener un output (que debemos poder guardar) del estilo:

9Er,5Vn8P$

Asegúrate de que se cumplan todos los criterios, y que estas contraseñas sean realmente aleatorias.

(EXTRA) Explora cómo podríamos hacer que la función copiara la contraseña automáticamente en el portapapeles del ordenador (como si la hubiéramos seleccionado y hecho ctrl+copy).

```python
# Se importan string (para poder obtener todos las letras, numeros y simbolos), random para poder generar contraseñas aleatorias y 
# pyperclip para poder copiar la contraseña al portapapeles.
import string
import random
try:
    import pyperclip
except ImportError:
    pyperclip = None
    
# esta funcion obtiene la longitud, mayusculas, minusculas, numeros y signos (en funcion de si estan activados con True o desactivados
# con false), crea un conjunto y se seleccionan mediante la condicion if, se crea el conjunto y se garantiza que al menos se seleccione uno,
# luego se rellena el resto de la longitud con caracteres aleatorios del conjunto, se mezclan con shuffle para que no esten siempre en el
# mismo lugar, se une la lista en un string, se copia la contraseña y se imprime el resultado.
def crear_contraseña(longitud, mayusculas=True, minusculas=True, numeros=True, signos=True, copiar_portapapeles=True):

    if not isinstance(longitud, int) or longitud <= 0:
        raise TypeError("La longitud debe ser un número entero positivo.")

    conjunto = ""
    obligatorios = []  

    if mayusculas:
        conjunto += string.ascii_uppercase
        obligatorios.append(random.choice(string.ascii_uppercase))

    if minusculas:
        conjunto += string.ascii_lowercase
        obligatorios.append(random.choice(string.ascii_lowercase))

    if numeros:
        conjunto += string.digits
        obligatorios.append(random.choice(string.digits))

    if signos:
        conjunto += string.punctuation
        obligatorios.append(random.choice(string.punctuation))

    if conjunto == "":
        raise ValueError("Debe activarse al menos un tipo de carácter.")

    if longitud < len(obligatorios):
        raise ValueError("La longitud no permite incluir todos los tipos seleccionados.")

    restantes = longitud - len(obligatorios)
    contrasena = obligatorios + [random.choice(conjunto) for _ in range(restantes)]

    random.shuffle(contrasena)

    contrasena = "".join(contrasena)

    if copiar_portapapeles:
        if pyperclip is not None:
            try:
                pyperclip.copy(contrasena)
                print("Contraseña copiada al portapapeles.")
            except Exception:
                print("No se pudo copiar al portapapeles.")
        else:
            print("pyperclip no está instalado.")

    return contrasena 

mi_contraseña = crear_contraseña(0, True, True, True, True, True)

print(f"Contraseña generada: {mi_contraseña}")
```

Una compañera de trabajo nos ha pedido un favor, aprovechando que sabe que estamos aprendiendo a programar. Tiene un histórico de partidos de fútbol catalán en un fichero, donde se almacenan los nombres de los equipos y los resultados. Necesita que procesemos los datos de forma automática, para extraer los resultados que necesita.

Utiliza el archivo "historic_partits.txt"

Necesita un programa que devuelva:

- El número total de goles que ha marcado cada equipo.

- El nombre del equipo más goleador.

- El nombre del equipo más goleado

- La clasificación global (cada victoria: 3 pts, empate 1 pts, derrota 0 pts)

Como referencia, los equipos con más puntos serían los siguientes:

![ejemplo ejercicio 2 nivel 3.png](ejemplo_ejercicio_2_nivel_3.png)

```python
import pandas as pd

# Se abre el .txt necesario (historic partits.txt) y se lee linea por linea
with open("historic partits.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()

# Diccionario para guardar datos de cada equipo
equipos = {}

# Se hace un for a líneas para saltar líneas vacías
for linea in lineas:
    linea = linea.strip()
    if linea == "":       
        continue

    # Se separa la línea en 3 partes (equipo local, marcador y equipo visitante para tener los 3 datos bien separados)
    partes = linea.split("\t")
    equipo_local    = partes[0]
    marcador        = partes[1]
    equipo_visitante = partes[2]

    # Se separa el marcador en goles
    goles = marcador.split("-")
    goles_local     = int(goles[0])
    goles_visitante = int(goles[1])

    # Se inicializa el equipo si aún no existe en el diccionario
    for equipo in [equipo_local, equipo_visitante]:
        if equipo not in equipos:
            equipos[equipo] = {"goles_marcados": 0, "goles_recibidos": 0, "puntos": 0}

    # Se crea el codigo para acumular goles marcados y recibidos en el diccionario
    equipos[equipo_local]["goles_marcados"]      += goles_local
    equipos[equipo_local]["goles_recibidos"]     += goles_visitante
    equipos[equipo_visitante]["goles_marcados"]  += goles_visitante
    equipos[equipo_visitante]["goles_recibidos"] += goles_local

    # Se calcula los puntos según el resultado, vicotira local, victoria visitante o empate respectivamente
    if goles_local > goles_visitante:       
        equipos[equipo_local]["puntos"] += 3
    elif goles_local < goles_visitante:     
        equipos[equipo_visitante]["puntos"] += 3
    else:                                   
        equipos[equipo_local]["puntos"] += 1
        equipos[equipo_visitante]["puntos"] += 1

# Se muestran los goles marcados por cada equipo, el equipo más goleador (el que más goles marcó), el equipo más goleado 
# (el que más goles recibió) y la clasificación ordenada por puntos

print("--- Goles marcados ---")
for equipo, estadistica in equipos.items():
    print(f"{equipo}: {estadistica['goles_marcados']}")

mas_goleador = max(equipos, key=lambda masGoles: equipos[masGoles]["goles_marcados"])
print(f"\nEquipo mas goleador: {mas_goleador} ({equipos[mas_goleador]['goles_marcados']} goles)")

mas_goleado = max(equipos, key=lambda masGoleado: equipos[masGoleado]["goles_recibidos"])
print(f"Equipo mas goleado: {mas_goleado} ({equipos[mas_goleado]['goles_recibidos']} goles recibidos)")
print()

# print("\n--- Clasificación ---")
# clasificacion = sorted(equipos.items(), key=lambda clasif: clasif[1]["puntos"], reverse=True)
# for n, (equipo, estadisticas) in enumerate(clasificacion, 1):
#     print(f"{n}. {equipo}: {estadisticas['puntos']} pts")

df = pd.DataFrame.from_dict(equipos, orient="index")
df = df.sort_values("puntos", ascending=False)
df.index.name = "Equipo"
df

```
